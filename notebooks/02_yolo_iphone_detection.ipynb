```python
import os
from ultralytics import YOLO
import torch
```

```python
print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("No GPU detected, training will run on CPU.")
```

```python
DATASET_YAML = r"C:\Users\user\Desktop\Smartphone_AI_Project\data\phone_classification\data.yaml"
print("Dataset YAML exists:", os.path.exists(DATASET_YAML))
```

```python
model = YOLO("yolov8n.pt")
print("YOLOv8n model loaded successfully.")
```

```python
results = model.train(
    data=DATASET_YAML,
    epochs=30,
    imgsz=640,
    batch=8,
    project=r"C:\Users\user\Desktop\Smartphone_AI_Project\models",
    name="iphone_yolo_stage2",
    device="cpu"
)
```

```python
metrics = model.val()
print(metrics)
```

```python
BEST_MODEL_PATH = r"C:\Users\user\Desktop\Smartphone_AI_Project\models\iphone_yolo_stage23\weights\best.pt"
print("Best model exists:", os.path.exists(BEST_MODEL_PATH))
```

```python
best_model = YOLO(BEST_MODEL_PATH)
print("Best trained YOLO model loaded.")
```

```python
TEST_FOLDER = r"C:\Users\user\Desktop\Smartphone_AI_Project\data\phone_classification\test\images"

results = best_model.predict(
    source=TEST_FOLDER,
    conf=0.25,
    save=True,
    show=False
)

print("Batch prediction complete.")
```
